```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme()

df = pd.read_csv("../data/raw/train.csv")
print(df.shape)
df[['Neighborhood','GrLivArea','BedroomAbvGr','FullBath','HalfBath','GarageCars','SalePrice']].isna().sum()
```

```python
miss = df.isna().sum().sort_values(ascending=False)
miss[miss > 0].plot(kind="barh", figsize=(8,10))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12,4))
sns.histplot(df["SalePrice"], kde=True, ax=ax[0]); ax[0].set_title("SalePrice")
sns.histplot(np.log1p(df["SalePrice"]), kde=True, ax=ax[1]); ax[1].set_title("log1p(SalePrice)")
```

```python
df[['GrLivArea','BedroomAbvGr','GarageCars','SalePrice']].corr()["SalePrice"].sort_values()
sns.scatterplot(data=df, x="GrLivArea", y="SalePrice")
```
